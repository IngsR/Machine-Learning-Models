## MD5

```python
import hashlib

password = "mypassword123"
md5_hash = hashlib.md5(password.encode()).hexdigest()
print("MD5 Hash:", md5_hash)
```

## Contoh Interaktif Hashing Dengan MD5

```python
import hashlib

# Input password dari user
password = input("Masukkan password untuk di-hash (MD5): ")

# Hash password dengan MD5
md5_hash = hashlib.md5(password.encode()).hexdigest()

# Tampilkan hasil hash
print("\n[MD5 Hash] ->", md5_hash)
```

## Verifikasi dengan MD5

```python
import hashlib

# Input password dari user
password = input("Masukkan password untuk di-hash (MD5): ")

# Hash password dengan MD5
md5_hash = hashlib.md5(password.encode()).hexdigest()

# Simpan hash ini (misalnya di database atau file)
print("\n[MD5 Hash] ->", md5_hash)

# --- Verifikasi Hash ---
verifikasi_password = input("\nMasukkan password lagi untuk verifikasi: ")
verifikasi_hash = hashlib.md5(verifikasi_password.encode()).hexdigest()

# Cek apakah hash cocok
if verifikasi_hash == md5_hash:
    print("✅ Password cocok! Autentikasi berhasil.")
else:
    print("❌ Password salah! Autentikasi gagal.")
```

## Bcrypt

```python
import bcrypt

salt = bcrypt.gensalt()
bcrypt_hash = bcrypt.hashpw(password.encode(), salt)
print("Bcrypt Hash:", bcrypt_hash)
```

## Contoh Interaktif Hashing Dengan Bcrypt

```python
import bcrypt

# Input password dari user
password = input("Masukkan password untuk di-hash (bcrypt): ").encode()

# Generate salt
salt = bcrypt.gensalt()

# Hash password dengan bcrypt
bcrypt_hash = bcrypt.hashpw(password, salt)

# Tampilkan hasil hash
print("\n[Bcrypt Hash] ->", bcrypt_hash.decode())
```

## Verifikasi dengan Bcrypt

```python
import bcrypt

# Input password dari user
password = input("Masukkan password untuk di-hash (Bcrypt): ").encode()

# Generate salt secara otomatis
salt = bcrypt.gensalt()

# Hash password dengan Bcrypt
bcrypt_hash = bcrypt.hashpw(password, salt)

# Simpan hash ini (misalnya di database)
print("\n[Bcrypt Hash] ->", bcrypt_hash.decode())

# --- Verifikasi Hash ---
verifikasi_password = input("\nMasukkan password lagi untuk verifikasi: ").encode()

# Cek apakah hash cocok
if bcrypt.checkpw(verifikasi_password, bcrypt_hash):
    print("✅ Password cocok! Autentikasi berhasil.")
else:
    print("❌ Password salah! Autentikasi gagal.")
```

## Contoh Interaktif Hashing Dengan Argon2

```python
from argon2 import PasswordHasher

# Inisialisasi Argon2 Password Hasher
ph = PasswordHasher()

# Input password dari user
password = input("Masukkan password untuk di-hash (Argon2): ")

# Hash password menggunakan Argon2
argon2_hash = ph.hash(password)

# Tampilkan hasil hash
print("\n[Argon2 Hash] ->", argon2_hash)
```

## Verifikasi dengan Argon2

```python
from argon2 import PasswordHasher

# Inisialisasi Argon2
ph = PasswordHasher()

# Input password dari user
password = input("Masukkan password untuk di-hash (Argon2): ")

# Hash password dengan Argon2
argon2_hash = ph.hash(password)

# Simpan hash ini (misalnya di database)
print("\n[Argon2 Hash] ->", argon2_hash)

# --- Verifikasi Hash ---
verifikasi_password = input("\nMasukkan password lagi untuk verifikasi: ")

try:
    if ph.verify(argon2_hash, verifikasi_password):
        print("✅ Password cocok! Autentikasi berhasil.")
except:
    print("❌ Password salah! Autentikasi gagal.")
```

## HMAC

```python
import hmac

key = b'secret_key'
hmac_hash = hmac.new(key, password.encode(), hashlib.sha256).hexdigest()
print("HMAC Hash:", hmac_hash)
```

## Contoh Interaktif Hashing Dengan HMAC

```python
import hmac
import hashlib

# Input password dan secret key
password = input("Masukkan password untuk di-hash (HMAC): ")
secret_key = input("Masukkan secret key: ")  # Kunci rahasia

# Buat hash menggunakan HMAC-SHA256
hmac_hash = hmac.new(secret_key.encode(), password.encode(), hashlib.sha256).hexdigest()

# Tampilkan hasil hash
print("\n[HMAC-SHA256 Hash] ->", hmac_hash)
```

### Verifikasi dengan HMAC-SHA256

```python
import hmac
import hashlib

# Kunci rahasia untuk HMAC (HARUS dirahasiakan!)
secret_key = b"my_super_secret_key"

# Input password dari user
password = input("Masukkan password untuk di-hash (HMAC-SHA256): ").encode()

# Hash password dengan HMAC-SHA256
hmac_hash = hmac.new(secret_key, password, hashlib.sha256).hexdigest()

# Simpan hash ini (misalnya di database)
print("\n[HMAC-SHA256 Hash] ->", hmac_hash)

# --- Verifikasi Hash ---
verifikasi_password = input("\nMasukkan password lagi untuk verifikasi: ").encode()

# Hash ulang dengan kunci yang sama
verifikasi_hash = hmac.new(secret_key, verifikasi_password, hashlib.sha256).hexdigest()

# Bandingkan hasil hash
if hmac.compare_digest(hmac_hash, verifikasi_hash):
    print("✅ Password cocok! Autentikasi berhasil.")
else:
    print("❌ Password salah! Autentikasi gagal.")
```
